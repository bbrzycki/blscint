# Screen Propagation: Perfect Narrowband Tone

This notebook treats an intrinsically perfect sine tone as a single-frequency field sample. The screen model does not simulate the carrier oscillation or an instrumental channelizer here; it computes the propagation gain at the tone frequency. A downstream voltage or spectrogram simulator can then decide how to represent the tone in samples or channels.

```python
from astropy import units as u
from blscint.simulations.screens import c95

source = c95.RadioSource(2e13 * u.cm)
screen = c95.Screen(
    distance=1e13 * u.cm,
    dx=5e5 * u.cm,
    dy=5e5 * u.cm,
    dz=1e12 * u.cm,
    shape=(160, 160),
    m_b2=30,
    seed=5,
)
model = c95.ScatteringModel(source, [screen], dx=5e5 * u.cm, dy=5e5 * u.cm, shape=(160, 160))

tone_field = model.observer_narrowband_tone_field(
    1 * u.GHz,
    v_trans=8e6 * u.cm / u.s,
)
tone_profile = tone_field.as_dynamic_spectrum(intrinsic_intensity=1.0, normalize="mean")

tone_profile.intensity.shape, tone_profile.frequencies, tone_profile.time_axis[:3], tone_field.phase[:3]
```

## Interpretation

For a perfect monochromatic tone, the screen layer can return both the complex propagated field and the derived scintillation gain profile at the tone frequency. The profile is a spatial cut through the observer plane, optionally converted to time with an assumed transverse velocity. The helper can sample a real passband voltage at a chosen observer point, but finite channel response, Doppler drift, receiver noise, and real voltage-pipeline effects still belong downstream.
