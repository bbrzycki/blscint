# Screen Propagation: Quantitative Validation

This notebook moves beyond visual behavior checks and asks whether the current screen implementations reproduce basic expectations from the papers.

Current scope:

- Coles-style phase-screen power-law slope and structure function.
- Strong-scattering intensity statistics and dynamic-spectrum decorrelation scales.
- Ravi & Deshpande-style dynamic-spectrum correlation behavior for matched and shifted signals.

These are still compact-grid validation runs, not final astrophysical reproductions.

## Coles-Style Phase Statistics

The radial phase power spectrum should follow the 2-D screen spectrum, approximately `q^-(alpha + 2)`.  For Kolmogorov-like `alpha = 5/3`, the target slope is `-11/3 = -3.67`.

The phase structure function should scale as `D_phi(s) ~ s^alpha`.  The current implementation follows the trend but is biased low in this compact validation setup, which is a sign that screen normalization/sampling/subharmonic treatment still needs work before claiming exact Coles reproduction.

![c95 phase validation](figures/validation_c95_phase_statistics.png)

Measured in the current demo run:

- Phase power-spectrum slope: `-3.59`, target `-3.67`.
- Phase structure-function slope: `1.48`, target `1.67`.

## Coles-Style Intensity Statistics

For strong scintillation, the observer-plane electric field should become close to complex Gaussian, so intensity gain should approach a unit-mean exponential distribution with variance near 1.  The dynamic-spectrum ACF gives empirical scintle decorrelation scales for this compact simulation.

![c95 intensity validation](figures/validation_c95_intensity_statistics.png)

Measured in the current demo run:

- Mean-normalized intensity variance: `0.96`, close to the exponential target of `1.0`.
- Exponential KS statistic: `0.007`.
- ACF e-folding scales: `1.44` time bins and `1.89` frequency bins.

## Ravi & Deshpande Correlation Check

The matched toy off-source spectrum contains `eta = 0.08` times the on-source scintillation pattern plus noise.  The shifted case uses the same pattern displaced in time/frequency, representing the kind of aliased or spatially displaced contaminant Ravi & Deshpande discuss.

![RD18 correlation validation](figures/validation_rd18_correlation.png)

Measured in the current demo run:

- Matched case: `eta_hat = 0.079`, peak lag `(0, 0)`.
- Shifted case: zero-lag `eta_hat = -0.001`, peak lag `(14, 9)`, matching the applied shift.
