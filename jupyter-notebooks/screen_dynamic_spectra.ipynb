## Minimal API Smoke Check

The new screen API returns an object with axes and metadata.  This cell keeps the shape and units visible.

```python
from astropy import units as u
from blscint.simulations.screens import c95

source = c95.RadioSource(2e13 * u.cm)
screen = c95.Screen(
    distance=1e13 * u.cm,
    dx=4e8 * u.cm,
    dy=4e8 * u.cm,
    dz=1e12 * u.cm,
    shape=(64, 64),
    m_b2=3.0,
    seed=10,
)
model = c95.ScatteringModel(source, [screen], dx=4e8 * u.cm, dy=4e8 * u.cm, shape=(64, 64))
ds = model.observer_dynamic_spectrum_result(
    1 * u.GHz,
    2 * u.MHz,
    16,
    v_trans=8e6 * u.cm / u.s,
    normalize="mean",
    progress=False,
)
ds.intensity.shape, ds.frequencies[[0, -1]], ds.time_axis[[0, -1]], ds.metadata
```
