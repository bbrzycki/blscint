## Direct Kernel Probe

This cell keeps the local API visible without depending on the plotting helper.

```python
import numpy as np
from astropy import units as u
from blscint.simulations.screens import hl07
from blscint.simulations.screens.base_classes import get_k

f = 1 * u.GHz
k = get_k(f).to(1 / u.cm)
q = np.array([0.0, 0.1, 0.25, 0.5]) * k
z = 5e4 * u.cm

fresnel = hl07.fresnel_transfer_function(q, z, f)
angular = hl07.angular_spectrum_transfer_function(q, z, f)
np.abs(angular - fresnel)
```
